## Introduction to Whisper

Whisper is a pre-trained deep learning model used for automatic speech recognition. It can take spoken audio as input and generate the corresponding text.

In this project, Whisper is used directly as a pre-trained model for inference instead of training a new speech recognition model from scratch.

## Why Whisper Was Selected

Whisper was selected because it is a pre-trained speech recognition model that is directly suitable for the Speech-to-Text application assigned in this project.

The model allows recorded audio to be provided as input and produces the recognized speech as text. This makes it suitable for demonstrating the use of a pre-trained deep learning model in a real-world application.

## Architecture / Working Principle

The Speech-to-Text application follows this workflow:

Audio Input
↓
Pre-trained Whisper Model
↓
Inference
↓
Speech Recognition
↓
Text Output

### Working Principle

1. The user provides a recorded audio file as input.
2. The audio is passed to the pre-trained Whisper model.
3. Whisper processes the audio and performs inference.
4. The model recognizes the spoken content.
5. The recognized speech is returned as text.
6. The transcribed text is displayed as the final output.

## Preprocessing

The uploaded audio file is provided directly to the Whisper `transcribe()` function.

Whisper internally prepares the audio for inference. It converts the audio into a log-Mel spectrogram, which is the representation used by the model. The audio is processed using 30-second windows, allowing the model to handle longer recordings as well.

Therefore, a separate manual preprocessing pipeline is not required in this implementation.

The preprocessing flow is:

Audio File
↓
Audio Loading
↓
Audio Preparation
↓
Log-Mel Spectrogram
↓
Whisper Model
↓
Transcribed Text

## Input Preparation

## Inference

Inference is the process of using the already-trained Whisper model to make predictions on new audio input.

In this project, the `transcribe()` function is used to perform inference on the selected audio file.

The following code performs the transcription:

result = model.transcribe(audio_file)

The recognized text is stored in the `result` variable.

## Output Processing

After the Whisper model completes inference, the transcribed speech is stored in the `text` field of the result.

The following code extracts and displays the recognized text:

print("Transcribed Text:")

## Limitations

1. Whisper may produce incorrect transcriptions when the audio contains background noise.
2. Accuracy can vary depending on the speaker's pronunciation and accent.
3. The base model requires more computational resources than very small speech-to-text models.
4. Poor-quality or unclear audio can reduce transcription accuracy.
5. The model may sometimes produce incorrect words when the speech is difficult to understand.
